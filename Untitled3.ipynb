```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout, GlobalAveragePooling2D
from tensorflow.keras.applications import MobileNetV2

base_model = MobileNetV2(
    weights='imagenet',
    include_top=False,
    input_shape=(224, 224, 3)
)

base_model.trainable = False

model = Sequential([
    base_model,
    GlobalAveragePooling2D(),
    Dense(128, activation='relu'),
    Dropout(0.5),
    Dense(4, activation='softmax')
])

model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
%%writefile gradcam.py
import tensorflow as tf
import numpy as np
import cv2

def make_gradcam_heatmap(img_array, model, last_conv_layer_name="out_relu"):

    base_model = model.layers[0]

    grad_model = tf.keras.models.Model(
        [base_model.input],
        [base_model.get_layer(last_conv_layer_name).output, model.output]
    )

    with tf.GradientTape() as tape:
        conv_outputs, predictions = grad_model(img_array)
        pred_index = tf.argmax(predictions[0])
        class_channel = predictions[:, pred_index]

    grads = tape.gradient(class_channel, conv_outputs)

    pooled_grads = tf.reduce_mean(grads, axis=(0, 1, 2))

    conv_outputs = conv_outputs[0]

    heatmap = conv_outputs @ pooled_grads[..., tf.newaxis]
    heatmap = tf.squeeze(heatmap)

    heatmap = tf.maximum(heatmap, 0)
    heatmap /= tf.math.reduce_max(heatmap)

    return heatmap.numpy()


def overlay_gradcam(img, heatmap):

    heatmap = cv2.resize(heatmap, (img.shape[1], img.shape[0]))
    heatmap = np.uint8(255 * heatmap)
    heatmap = cv2.applyColorMap(heatmap, cv2.COLORMAP_JET)

    return cv2.addWeighted(img, 0.6, heatmap, 0.4, 0)
```

```python
%%writefile segmentation.py
import cv2
import numpy as np

def segment_eye(img):

    gray = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)

    blur = cv2.GaussianBlur(gray, (5,5), 0)

    _, thresh = cv2.threshold(blur, 70, 255, cv2.THRESH_BINARY)

    return thresh
```

```python
import streamlit as st
import tensorflow as tf
import numpy as np
import cv2
from PIL import Image
import time
import psutil

from gradcam import make_gradcam_heatmap, overlay_gradcam
from segmentation import segment_eye

# =====================
# LOAD MODEL
# =====================
@st.cache_resource
def load_model():
    return tf.keras.models.load_model("model.h5")

model = load_model()

class_names = ['Cataract', 'Jaundice', 'Normal_Eye', 'Pterygium']

descriptions = {
    "Cataract": "Katarak adalah kekeruhan lensa mata yang menyebabkan penglihatan kabur.",
    "Jaundice": "Mata menguning akibat peningkatan bilirubin.",
    "Normal_Eye": "Mata normal tanpa indikasi penyakit.",
    "Pterygium": "Pertumbuhan jaringan pada permukaan mata."
}

# =====================
# PREDICT FUNCTION
# =====================
def predict(image):
    img = image.resize((224, 224))
    img_array = np.array(img) / 255.0
    img_array = np.expand_dims(img_array, axis=0)

    start = time.time()
    pred = model.predict(img_array)
    end = time.time()

    idx = np.argmax(pred)
    confidence = np.max(pred)

    ram = psutil.virtual_memory().used / (1024**3)
    inference_time = end - start

    return idx, confidence, pred, img_array, ram, inference_time

# =====================
# UI
# =====================
st.title("🧠 Eye Disease AI System")
st.write("Upload gambar mata untuk deteksi penyakit.")

uploaded_file = st.file_uploader("Upload Image", type=["jpg", "png", "jpeg"])

if uploaded_file:

    image = Image.open(uploaded_file).convert("RGB")
    st.image(image, caption="Input Image", use_container_width=True)

    if st.button("Predict"):

        idx, conf, pred, img_array, ram, t = predict(image)
        label = class_names[idx]

        st.success(f"Hasil: {label}")
        st.info(f"Confidence: {conf*100:.2f}%")

        st.write("### Probabilitas")
        for i, c in enumerate(class_names):
            st.write(f"{c}: {pred[0][i]*100:.2f}%")

        st.write("### Penjelasan")
        st.write(descriptions[label])

        st.write("### Evaluasi Runtime")
        st.write(f"Inference Time: {t:.4f} sec")
        st.write(f"RAM Usage: {ram:.2f} GB")

        # Segmentasi
        st.write("### Segmentasi")
        seg = segment_eye(np.array(image))
        st.image(seg, use_container_width=True)

        # GradCAM
        st.write("### Grad-CAM")
        heatmap = make_gradcam_heatmap(img_array, model)
        gradcam = overlay_gradcam(np.array(image), heatmap)
        st.image(gradcam, use_container_width=True)
```

```python
import tensorflow as tf
import numpy as np
import cv2

def make_gradcam_heatmap(img_array, model, last_conv_layer_name="out_relu"):
    base_model = model.layers[0]

    grad_model = tf.keras.models.Model(
        [base_model.input],
        [base_model.get_layer(last_conv_layer_name).output, model.output]
    )

    with tf.GradientTape() as tape:
        conv, pred = grad_model(img_array)
        class_idx = tf.argmax(pred[0])
        loss = pred[:, class_idx]

    grads = tape.gradient(loss, conv)
    pooled = tf.reduce_mean(grads, axis=(0,1,2))

    conv = conv[0]
    heatmap = conv @ pooled[..., tf.newaxis]
    heatmap = tf.squeeze(heatmap)

    heatmap = tf.maximum(heatmap, 0)
    heatmap /= tf.math.reduce_max(heatmap)

    return heatmap.numpy()


def overlay_gradcam(img, heatmap):
    heatmap = cv2.resize(heatmap, (img.shape[1], img.shape[0]))
    heatmap = np.uint8(255 * heatmap)
    heatmap = cv2.applyColorMap(heatmap, cv2.COLORMAP_JET)

    return cv2.addWeighted(img, 0.6, heatmap, 0.4, 0)
```

```python
import streamlit as st
import numpy as np
import pandas as pd
import os
from PIL import Image
from datetime import datetime

# =========================
# PAGE CONFIG
# =========================
st.set_page_config(
    page_title="AI Medical Eye System",
    page_icon="🧠",
    layout="wide"
)

# =========================
# MOCK MODEL (SAFE FOR HF)
# =========================
class_names = ['Cataract', 'Jaundice', 'Normal_Eye', 'Pterygium']

def fake_predict(img):
    # simulasi prediction (biar tidak crash di HF)
    probs = np.random.dirichlet(np.ones(len(class_names)))
    idx = np.argmax(probs)
    return idx, probs[idx], probs

# =========================
# HISTORY (SAFE VERSION)
# =========================
HISTORY_FILE = "patient_history.csv"

if not os.path.exists(HISTORY_FILE):
    pd.DataFrame(columns=[
        "Time", "Name", "Age", "Gender",
        "Diagnosis", "Confidence"
    ]).to_csv(HISTORY_FILE, index=False)

def save_history(name, age, gender, diagnosis, confidence):
    new_data = pd.DataFrame([{
        "Time": datetime.now().strftime("%Y-%m-%d %H:%M:%S"),
        "Name": name,
        "Age": age,
        "Gender": gender,
        "Diagnosis": diagnosis,
        "Confidence": float(confidence)
    }])

    df = pd.read_csv(HISTORY_FILE)
    df = pd.concat([df, new_data], ignore_index=True)
    df.to_csv(HISTORY_FILE, index=False)

# =========================
# UI
# =========================
st.title("🧠 AI Medical Eye Disease System (Demo)")
st.write("Upload gambar mata untuk simulasi diagnosis AI")

uploaded_file = st.file_uploader("Upload Eye Image", type=["jpg", "png", "jpeg"])

patient_name = st.text_input("Name")
patient_age = st.number_input("Age", 0, 120, 25)
patient_gender = st.selectbox("Gender", ["Male", "Female"])

# =========================
# MAIN
# =========================
if uploaded_file:

    image = Image.open(uploaded_file).convert("RGB")
    st.image(image, caption="Input Image", use_container_width=True)

    if st.button("🔍 Run AI Diagnosis"):

        idx, conf, pred = fake_predict(image)
        label = class_names[idx]

        status = "🟢 Normal" if label == "Normal_Eye" else "🔴 Abnormal"

        save_history(patient_name, patient_age, patient_gender, label, conf)

        st.subheader("🧠 Result")
        st.success(label)
        st.info(f"Confidence: {conf*100:.2f}%")
        st.warning(status)

        df_pred = pd.DataFrame({
            "Class": class_names,
            "Probability": pred
        })

        st.bar_chart(df_pred.set_index("Class"))

        report = f"""
Name: {patient_name}
Age: {patient_age}
Gender: {patient_gender}
Diagnosis: {label}
Confidence: {conf*100:.2f}%
Time: {datetime.now()}
"""

        st.download_button("Download Report", report, file_name="report.txt")

# =========================
# HISTORY
# =========================
st.markdown("---")
st.subheader("History")

history_df = pd.read_csv(HISTORY_FILE)
st.dataframe(history_df)
```
